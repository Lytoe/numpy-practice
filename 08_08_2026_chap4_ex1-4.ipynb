```python
# we're gonna define cosine similarity :
v1 = np.array([15,8])
v2 = np.array([3,9])

a = np.dot(v1,v2)
mag_v1 = np.linalg.norm(v1)
mag_v2 = np.linalg.norm(v2)


deno = mag_v1 * mag_v2 
thewhole = a / deno
print(thewhole)
core = np.corrcoef(v1,v2)
print(core)

```
