```python
def pearson_correlation(v1, v2):
    # Step 1: Mean-center
    mean1 = np.mean(v1)
    mean2 = np.mean(v2)
    
    centered1 = [x - mean1 for x in v1]
    centered2 = [y - mean2 for y in v2]
    
    # Step 2: Calculate numerator
    numerator = 0
    for i in range(len(v1)):
        numerator += centered1[i] * centered2[i]
    
    # Step 3: Calculate denominator
    sum_sq1 = 0
    sum_sq2 = 0
    for i in range(len(v1)):
        sum_sq1 += centered1[i] ** 2
        sum_sq2 += centered2[i] ** 2
    
    denominator = np.sqrt(sum_sq1 * sum_sq2)
    
    # Step 4: Return correlation
    return numerator / denominator
```
