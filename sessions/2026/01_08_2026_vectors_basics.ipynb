```python
import numpy as np
import matplotlib.pyplot as plt
import random

rand_list_a = [random.randint(-4,4) for _ in range(100)]
rand_list_b = [random.randint(-4,4) for _ in range(100)]
the_first = np.array([3, 5, 1])
the_second = np.array([0, 2, 2])
new_vecs = []

for i in range(100):
    point = rand_list_a[i] * the_first + rand_list_b[i] * the_second
    new_vecs.append(point)
new_vecs = np.array(new_vecs)


import plotly.express as px
import pandas as pd

# Create dataframe
df = pd.DataFrame(new_vecs, columns=['x', 'y','z'])

# ONE LINE PLOT!
fig = px.scatter_3d(df, x='x', y='y', z='z', title='Random points in subspace')
fig.show()
```
