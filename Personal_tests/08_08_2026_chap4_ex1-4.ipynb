```python
import pandas as pd
import numpy as np
import plotly.express as px
import matplotlib as plt
import math

"""today I wanted to try filtering signals with simple dotproduct.
a simple function slides through the the array of numbers with one number intentionally devoted to act
as anomoly, it detects spikes(anomoly)
"""
```

```python
signal = np.array([1,2,-2,3,2,5,1,6])
arr = np.random.random(20)
print(arr)
arr[10] = 13
print(arr)

signal = arr
output = []
"""Slide kernel over signal, computing dot product at each position."""

def slidin_ker (signal, kernel):
    signal_len = len(signal)
    kernel_len =len(kernel)
    output_len = signal_len - kernel_len +1

    for i in range(output_len):
        window = signal[i:i+kernel_len]
        dotprod = np.dot(window,kernel)

        output.append(dotprod)

    return np.round(output).astype(int) 
```

```python
ker = np.array([-2,3,-2])
slidin_ker(arr,ker)
```
