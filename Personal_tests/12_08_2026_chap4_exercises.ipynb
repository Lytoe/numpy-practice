```python
import numpy as np
```

```python
def signal_prod():
    kernel = np.array([1,-1])
    base_signal = np.full(30, -1,dtype=float)
    base_signal[10:20] = 1.5
    len_base = len(base_signal)
    len_ker = len(kernel)
    edge_series = np.zeros(len_base - len_ker + 1)
    for i in range(len_base - len_ker + 1):
        w =base_signal[i:i+len_ker]
        
        edge_series[i] = np.dot(kernel,w)
        print()
    return(edge_series)


  

```
