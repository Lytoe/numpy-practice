```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp('I can promise it is worth you time')

sent = ''
for i, token in enumerate(doc):
    
    if token.tag_ == 'PRP' and doc[i+1].tag_ == "MD" and doc[i+2] == "VB":
        sent = doc[i+1].text.capitalize() + '' + doc[i].text
        sent = sent + ' ' + doc[i+2:].text
        print(sent)
        break
        
```
