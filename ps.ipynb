```python
import pandas as pd
import matplotlib as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv("AHD.csv")
df.shape
df.head()
df.isna().any().any()
```

```python
df.isna().any()
```

```python
df.head()
```

```python
missing  = df.isnull().mean()
print(missing)
cols_to_drop = missing[missing >= 0.30]
print(cols_to_drop.keys())
type(cols_to_drop)
```

```python
df = df.drop(columns=cols_to_drop.keys())
df.head()

```

```python
sns.histplot(df["SalePrice"])
```

```python
df_numerical = df.select_dtypes(include=['number'])
print(df_numerical)

for col in df_numerical.columns:
      plt.figure()
      plt.title(col)
      df_numerical[col].hist()

```
