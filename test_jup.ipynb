```python
import numpy as np
v = np.array([1,1])

print("norm of A :" ,np.linalg.norm(v))

```

```python
hh =[]

for element in v:
    hh.append(element**2)

the_norm = math.sqrt(sum(hh))

print(the_norm)
```

```python
v_unit = v / the_norm
print(v_unit)
```

```python
def unit_maker(vector):
    norm = np.linalg.norm(vector)
    unite = vector / norm
    print(unite.shape)
    return unite
```

```python
new_vector = np.array([[10,347,4],[15,475,225252526],[10,4585,9]])
unit_maker(new_vector)
```
